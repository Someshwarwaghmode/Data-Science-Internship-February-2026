## Problem 1: Smart Light Controller
### Problem Statement
#### Create a class to control a smart light that can be turned ON or OFF and display its status.


```python
class smart_light:
    def __init__(self,light_name):
        self.light_name = light_name
        self.status = "OFF"

    def trun_on(self):
        self.status = "ON"

    def trun_off(self):
        self.status = "OFF"

    def display_status(self):
        print()
        print(f"{self.light_name} is {self.status}")

light_name = input("Enter The name of light::")
status = input("Enter the Status::").upper()

light = smart_light(light_name)

if status == "ON":
    light.trun_on()
    light.display_status()

elif status == "OFF":
    light.trun_off()
    light.display_status()
else:
    print("INvalid Input")
```

## Problem 2: Employee ID Card System
### Problem Statement
#### Create a class that stores employee name, ID, and department and displays ID card details.


```python
class Employee_id_card:
    def __init__(self, name, emp_id, department):
        self.name = name
        self.emp_id = emp_id
        self.department = department 

    def id_card(self):
        print()
        print("ID CARD DETAILES")
        print(f"Employee Name:   {self.name}")
        print(f"Employee ID:     {self.emp_id}")
        print(f"Department Name: {self.department}")


name = input("Enter Employee Name::")
emp_id = input("Enter Employee Id::")
department = input("Enter Department Name::")
        
emp_id = Employee_id_card(name, emp_id, department)
emp_id.id_card()
```

## Problem 3: Mobile Contact Record
### Problem Statement
#### Create a class to store contact name and phone number and display contact information


```python
class Contacts:
    def __init__(self,name,phone):
        self.name = name
        self.phone = phone
    def display_contact(self):
        print()
        print("     Contact Saved     ")
        print(f"Name  : {self.name}")
        print(f"Phone : {self.phone}")

name = input("Enter Name::")
phone = input("Enter Phone::")

contact1 = Contacts(name,phone)

contact1.display_contact()
```

## Problem 4: Product Price Tag Generator
### Problem Statement
#### Create a class that stores product name and price and prints a formatted price tag.


```python
class Product_details:
    def __init__(self, p_name, p_price):
        self.p_name = p_name
        self.p_price = p_price

    def display_detailse(self):
        print()
        print(f"Product: {self.p_name}")
        print(f"Price  : \u20B9{self.p_price}")

product = input("Enter Product Name:: ")
price = int(input("Enter Price:: "))

product1 = Product_details(product,price)

product1.display_detailse()
        
```

## Problem 5: Movie Rating Display System
### Problem Statement
#### Create a class that stores movie name and rating and displays movie details


```python
class Movie_Store:
    def __init__(self, movie_name, rate):
        self.movie_name = movie_name
        self.rate = rate

    def display_movie_rating(self):
        print()
        print(f"Movie: {self.movie_name}")
        print(f"Rate : {self.rate}")

#input
movie_name = input("Enter Movie Name:: ").lower()
rate = float(input("Enter Rating:: "))

#obj creation
movie1 = Movie_Store(movie_name, rate)

#calling function
movie1.display_movie_rating()
```
