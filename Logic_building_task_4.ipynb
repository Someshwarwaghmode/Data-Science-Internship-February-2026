## Mobile Recharge Validation System

```python
def valid_recharge(amt):
    while True:
        if amt > 50 and amt in [199, 299, 399, 599]:
            return f"Recharge Successful: {amt}"
        elif amt <= 50:
            amt = int(input("Invalid Amount! Enter ( > 50 ): "))
        else:
            amt = int(input("Invalid Plan Enter Valide [199, 299, 399, 599]: "))
        
                  
```

```python
input_amt = int(input("Enter Amount:: "))
result = valid_recharge(input_amt)
print(result)
```

## Inventory Reorder Alert System

```python
def alert_system(product_info):
    for product, quantity in product_info.items():
        if quantity < 15:
            print(f"{product}: Reorder alert (stock: {quantity})")
        else:
            print(f"{product}: Ok (stock: {quantity})")
        
    
```

```python
product_info = {
    "Apples": 50,
    "Oranges": 30,
    "Milk_Cartons": 15,
    "Bread_Loaves": 10,
    "banana":6,
    "milk":13
}
alert_system(product_info)
```
