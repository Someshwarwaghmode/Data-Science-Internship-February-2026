## Mobile Recharge Validation System

```python
def valid_recharge(amt):
    while True:
        if amt > 50 and amt in [199, 299, 399, 599]:
            return f"Recharge Successful: {amt}"
        elif amt <= 50:
            amt = int(input("Invalid Amount! Enter ( > 50 ): "))
        else:
            amt = int(input("Invalid Plan Enter Valide [199, 299, 399, 599]: "))
        
                  
```

```python
input_amt = int(input("Enter Amount:: "))
result = valid_recharge(input_amt)
print(result)
```

## Inventory Reorder Alert System

```python
def alert_system(product_info):
    for product, quantity in product_info.items():
        if quantity < 15:
            print(f"{product}: Reorder alert (stock: {quantity})")
        else:
            print(f"{product}: Ok (stock: {quantity})")
        
    
```

```python
product_info = {
    "Apples": 50,
    "Oranges": 30,
    "Milk_Cartons": 15,
    "Bread_Loaves": 10,
    "banana":6,
    "milk":13
}
alert_system(product_info)
```

## Student Result Processing System

```python
def mark_evalution_system(marks):
    total = 0
    for mark in marks:
        total += mark
    avg = total/len(marks)
    if avg >= 50:
        return f"Pass with Avg mark: {avg}"
    else:
        return f"Fail with Avg mark: {avg}"
```

```python
marks = [65,54,65,21,55]
result = mark_evalution_system(marks)
print(result)
```

```python
marks = [24,35,27,22,21]
result = mark_evalution_system(marks)
print(result)
```

## Cab Fare Estimator with Retry Option

```python
def fare_calculation(distance, peak):
    base_fare = 50
    per_km_rate = 12
    fare = base_fare + (distance * per_km_rate)
    if(peak == "yes"):
        fare = fare+(fare * 0.25)
    return fare
    
while True:
    distance = int(input("Enter Distance in Km: "))
    peak = input("Is it peak hour? (yes/no): ").lower()
    total_fare = fare_calculation(distance, peak)
    print(f"Total Fare: {total_fare} rupees")
    retry = input("DO u Want to calculate again? (yes/no): ").lower()
    if retry != "yes":
        break
        
```

## Employee Attendance Eligibility 

```python
def calculate_attendance(attendance_sheet):

    present_days = 0;
    for P in attendance_sheet:
        if P == "P":
            present_days += 1

    percentage = (present_days/len(attendance_sheet))*100
    if percentage >= 75:
        return f"Eligible: {percentage}%"
    else:
        return f"Not Eligible: {percentage}%"
    
```

```python
list = ["P", "A","P","P"]
result = calculate_attendance(list)
print(result)
```
