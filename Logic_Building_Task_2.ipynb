# Problem 1: Unique Words in a Sentence

```python
input_string = "python is easy and python is powerful"

list_of_world = input_string.split()
list_to_set = set(list_of_world)

print(f"Unique world count: {len(list_to_set)}")
print(f"Unique world: {list_to_set}")
    
```

# Problem 2: Highest Salary from Employee Data

```python
employees = {
    "Ravi":75000,
    "Anita":68000,
    "Kiran":72000
}
#Get name of employee who's salary is High
employee_name = max(employees, key = employees.get)

print(f"Highest Salary: {employee_name} - {employees[employee_name]}")
    
```

# Problem 3: Find Maximum and Minimum Values

```python
input_numbers = [45,22,89,10,66]

max_num = input_numbers[0]
min_num = input_numbers[0]

for num in input_numbers:
    max_num = max(max_num, num)
    min_num = min(min_num, num)
    
print(f"List: {input_numbers}")
print(f"Max: {max_num}")
print(f"Min: {min_num}")
```

# Problem 4: Count Products Above a Price Threshold

```python
input_prices = [450,1200,899,1500,300]

count = 0
for price in input_prices:
    if price > 1000:
        count += 1

print(f"Product above 1000: {count}")
```

# Problem 5: Calculate Attendance Percentage

```python
input_attendance = ["P","P","A","P","P"]

present_count = 0
for present in input_attendance:
    if present == "P":
        present_count += 1

percentage = (present_count/len(input_attendance))*100

print(F"Attendance Percentage: {percentage}")
```

# Problem 7: Count Character Frequency

```python
input_string = "ppython"

count_dic = dict()
Set = set()
for char in input_string:
    if char not in Set:
        Set.add(char)
        count_dic[char] = 1
    else:
        count_dic[char] = count_dic[char]+1

print(count_dic)
```

# Problem 9: Check if a Key Exists in a Dictionary

```python
employees_dict = {
    "Ravi":75000,
    "Anita":68000,
    "Kiran":72000
}

employee_name= input("Enter Employee name: ")

if employee_name in employees_dict.keys():
    print("Employee exists")
else:
    print("Employee not exists")
```

# Problem 10: Calculate Average Marks

```python
subject_marks = [75,80,70,60,77.5]

total_marks = 0

for mark in subject_marks:
    total_marks = total_marks+mark

total_subjects = len(subject_marks)
avg_mark = (total_marks/total_subjects)

print(f"Average Marks: {avg_mark}")
```
