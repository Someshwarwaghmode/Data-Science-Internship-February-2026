3. Two real world scenarios are sentiment analysis and Q/A chatbot:-

- In sentiment analysis, removing stopwords can change the meaning of a sentence because words like “not”, “no”, and “never” are important. For example, “This movie is not good” may become “movie good,” which gives the wrong (positive) sentiment.

 - In question answering or chatbots, stopwords help maintain the structure and intent of a sentence. For example, “What is the time now?” becomes “time now” after removal, making the query unclear and harder to understand.

4. Stemming and lemmatization are both used to convert words into their base form, but they work differently. Stemming is a simple process that just cuts off prefixes or suffixes, so it can sometimes produce incorrect or incomplete words, like “studies” becoming “studi.” On the other hand, lemmatization is more accurate because it understands the meaning of the word and converts it into a proper root form, such as “studies” becoming “study.”

```python
import re
from collections import Counter
```

```python
def preprocessing_text(text):
    #1. handling error
    if not text or not isinstance(text, str):
        return [], ""


    # 2. uppercase ---> lowercase
    text = text.lower()
    
    # 3. remove URL
    text = re.sub(r'http\S+|www\S+', '', text)

    # 4. remove email
    text = re.sub(r'\S+@\S+', '', text)

    # 5. remove number
    text = re.sub(r'\d+', '',  text)

    # 6. remove emojis and special characters
    text = re.sub(r'[^\w\s]', '', text)

    # 7. handle repeated character (soooooo --> soo)
    text = re.sub(r'(.)\1{2,}', r'\1\1', text)

    # 8. remove extra spaces 
    text = re.sub(r'\s+', ' ', text).strip()

    # 9. tokenization
    tokens = text.split()

    # 10.Remove short word ( <= 2 ) keep no, not
    important_words = {'no', 'not'}
    tokens = [word for word in tokens if len(word) > 2 or word in important_words]

    clean_sentence = " ".join(tokens)

    return tokens, clean_sentence
```

```python
test_sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely loooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]
```

```python
#task 3
for sentence in test_sentences:
    tokens, clean = preprocessing_text(sentence)

    print("\nOrginal: ", sentence)
    print("Tokens: ", tokens)
    print("Clean: ", clean)
```

```python
# task 4
def token_analysis(tokens):
    total = len(tokens)
    unique = len(set(tokens))
    avg_len = sum(len(word) for word in tokens) / total if total > 0 else 0

    return total, unique, avg_len

print("\n ======== token Analysis===========")
for sentence in test_sentences:
    tokens, clean = preprocessing_text(sentence)
    total, unique, avg_len = token_analysis(tokens)

    print("\nSnetence:", clean)
    print("Total tokens:", total)
    print("Unique tokens:", unique)
    print("Avg, token length:", round(avg_len, 2))
```

```python
# task 5

all_tokens = []

for sentence in test_sentences:
    tokens, _ = preprocessing_text(sentence)
    all_tokens.extend(tokens)

    counter = Counter(all_tokens)

print("\n ======== frequency analysis =========")

print("TOp 10 frequent words:")
print(counter.most_common(10))

print("\n Leat 5 frequent words")
print(counter.most_common()[-5:])
```

```python
# task 6
def full_pipeline(text_list):
    all_tokens = []
    clean_sentences = []

    for text in text_list:
        tokens, clean = preprocessing_text(text)
        all_tokens.append(tokens)
        clean_sentences.append(clean)

    return{
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }


print("\n ========FULL Pipeline==============")

result = full_pipeline(test_sentences)

print("\nTokens:")
for t in result["tokens"]:
    print(t)


print("\nClean Sentences:")
for s in result["clean_sentences"]:
    print(s)      
```
