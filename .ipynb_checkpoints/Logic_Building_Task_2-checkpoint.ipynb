# Problem 1: Unique Words in a Sentence

```python
input_string = "python is easy and python is powerful"

#here is Split String and Convert it into set to make all Unique Words 
convert_to_set = set(input_string.split())

#Print Count and Unique Words of input_string
print(f"Unique words count : {len(convert_to_set)}")
print(f"Unique words : {convert_to_set}")
    
```

# Problem 2: Highest Salary from Employee Data

```python
employees = {
    "Ravi":75000,
    "Anita":68000,
    "Kiran":72000
}
#Get name of employee who's salary is High
employee_name = max(employees, key = employees.get)
print(f"Highest Salary: {employee_name} - {employees[employee_name]}")
    
```
