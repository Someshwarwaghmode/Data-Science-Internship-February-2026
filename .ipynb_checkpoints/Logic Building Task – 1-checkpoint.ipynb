1. User Login Check

```python
#User Login Check Function
def Login(username, password):
    if(username == "admin" and password == 1234):
        print("Login Succseful")
    else:
        print("Invalid Credentials")
```

```python
#function Call with Valid Username Name and Password
Login("admin",1234)

#Function Call With Invalid Username But Valid Password
Login("student",1234)

#Function Call With valid Username and Invalid Password
Login("admin",4321)
```

2. Pass / Fail Analyzer

```python
#Function for Academic evaluation systems
def Evaluation(marks):
    pass_count = 0
    fail_count = 0
    for mark in marks:
        if(mark >= 50):
            pass_count +=1
        else:
            fail_count +=1
    print("Total Student Count = ",len(marks))
    print("Pass Student Count = ",pass_count)
    print("Fail Student Count = ",fail_count)
```

```python
#Here Call The Evalution Function
marks = [45, 78, 90, 33, 60]
Evaluation(marks)
```

3. Simple Data Cleaner

```python
#Function For Cleaning data
def dataCleaning(names):
    index = 0;
    for name in names:
        names[index] = name.strip().lower()
        index +=1
    print("Cleaned Name List::",names)
```

```python
#Here Calling the Function for Clean the Data
names = [" Alice ", "bob", " CHARLIE "]
dataCleaning(names)
```

4. Message Length Analyzer

```python
#function for the length Analyzer 
def lengthAnalyzer(messages):
    for message in messages:
        lenght = len(message)
        print("Message: ",message)
        print("Message Count: ",lenght)

        if(lenght >= 10):
            print("Flag : Message Longer Than 10 Char")
            
        print()
```

```python
#here Calling the function
messages = ["Hi", "Welcome to the platform", "OK"]
lengthAnalyzer(messages)
```

5. Error Message Detector

```python
#function for the Count Error
def errorEntries(logs):
    error_count = 0
    for log in logs:
        if(log == "ERROR"):
            error_count +=1
    print("Total Error Entries: ",error_count)
        
    
```

```python
logs = ["INFO", "ERROR", "WARNING", "ERROR"]
errorEntries(logs)
```
