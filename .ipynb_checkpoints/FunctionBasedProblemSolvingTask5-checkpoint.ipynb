## Post Engagement Analyzer

### Create a function to analyze engagement on a social media post.
### Rules
#### Calculate total likes
#### If Total Likes ≥ 1000 → Viral Post
#### Otherwise → Normal Engagement


```python
# function for Engagement Analyzer
def engagement_analyzer(likes):

    total_like = 0
    for like in likes:
        total_like += like

    status = ""
    if total_like >= 1000:
        status = "Viral Post"
    else:
        status = "Normal Engagement"
    return status,total_like

# Function call
likes = [45,55,65,44,66]
result,total = engagement_analyzer(likes)
print(f"total Likes: {total}")
print(f"Status: {result}")
```

# Medicine Stock Alert System

### Create a function to monitor medicine stock levels.
### Rules
#### Stock < 10 → Low Stock Alert
#### Stock ≥ 10 → Stock Sufficient

```python
## function for Medicion stock monitor
def medicine_stock_monitor(stock):
    if stock < 10:
        stock = "Low Stock Alert"
    else:
        stock = "Stock Sufficient"

    return stock

# Function call
stock = 68
status = medicine_stock_monitor(stock)
print(f"Medicine Stock: {stock}")
print(f"Status: {status}")
```

# Rainfall Adequacy Checker

### Create a function that analyzes seasonal rainfall data.
### Rules
#### Calculate average rainfall
#### Average ≥ required level → Adequate Rainfall
#### Otherwise → Inadequate Rainfall

```python
def rainfall_checker(rainfall, required):
    average = sum(rainfall) / len(rainfall)
    print("Average Rainfall:", average)

    if average >= required:
        print("Rainfall Status: Adequate Rainfall")
    else:
        print("Rainfall Status: Inadequate Rainfall")

rainfall_data = [70, 75, 68, 80, 67]
rainfall_checker(rainfall_data, 70)
```

## Duplicate Account Detection
### Problem Statement
#### Create a function to detect duplicate usernames in a system.
### Rules
#### If duplicates exist → Flag accounts
#### Else → Accounts are unique


```python
def check_duplicates(usernames):
    if len(usernames) != len(set(usernames)):
        print("Duplicate Accounts Found: Yes")
    else:
        print("Duplicate Accounts Found: No")

users = ["abhi", "rahul", "abhi", "neha"]
check_duplicates(users)
```

## Premium Crop Price Filter
### Problem Statement
#### Create a function that filters crop prices above a premium threshold.
### Rules
#### Price > ₹2000 → Premium Crop
#### Else → Regular Crop


```python
def premium_crops(prices):
    premium = [price for price in prices if price > 2000]
    print("Premium Crops:", premium)

crop_prices = [1500, 2500, 1800, 3200]
premium_crops(crop_prices)
     
```

## System Monitoring – Application Health Checker
### Problem Statement
#### Create a function to evaluate application health based on error count.
### Rules
#### Errors = 0 → Healthy
#### Errors ≤ 5 → Minor Issues
#### Errors > 5 → Critical Issues

```python
def system_health(errors):
    print("Error Count:", errors)

    if errors == 0:
        print("System Status: Healthy")
    elif errors <= 5:
        print("System Status: Minor Issues")
    else:
        print("System Status: Critical Issues")

system_health(7)
```

## Student Attendance Eligibility System
### Problem Statement
#### Create a function to calculate attendance percentage and check exam eligibility.
### Rules
#### Attendance ≥ 75% → Eligible
#### Attendance < 75% → Not Eligible


```python
def attendance_checker(attendance):
    present = attendance.count("P")
    percentage = (present / len(attendance)) * 100

    print("Attendance Percentage:", percentage)

    if percentage >= 75:
        print("Exam Eligibility: Eligible")
    else:
        print("Exam Eligibility: Not Eligible")

attendance = ["P", "P", "P", "P", "A"]
attendance_checker(attendance)
```

## Smart Electricity Bill Analyzer
### Problem Statement
#### You are building a smart electricity billing system for residential users.
#### Create a function that calculates the electricity bill and classifies usage level based on units consumed.
### Billing Rules
#### First 100 units → ₹3 per unit
#### Next 100 units (101–200) → ₹5 per unit
#### Above 200 units → ₹7 per unit


```python
def electricity_bill(units):
    if units <= 100:
        bill = units * 3
    elif units <= 200:
        bill = (100 * 3) + ((units - 100) * 5)
    else:
        bill = (100 * 3) + (100 * 5) + ((units - 200) * 7)

    if bill < 500:
        status = "Low Usage"
    elif bill <= 1500:
        status = "Moderate Usage"
    else:
        status = "High Usage"

    return bill, status


bill, status = electricity_bill(250)
print("Total Bill:", bill)
print("Usage Level:", status)
```
