## Post Engagement Analyzer

### Create a function to analyze engagement on a social media post.
### Rules
#### Calculate total likes
#### If Total Likes ≥ 1000 → Viral Post
#### Otherwise → Normal Engagement


```python
# function for Engagement Analyzer
def engagement_analyzer(likes):

    total_like = 0
    for like in likes:
        total_like += like

    status = ""
    if total_like >= 1000:
        status = "Viral Post"
    else:
        status = "Normal Engagement"
    return status,total_like

# Function call
likes = [45,55,65,44,66]
result,total = engagement_analyzer(likes)
print(f"total Likes: {total}")
print(f"Status: {result}")
```

# Medicine Stock Alert System

### Create a function to monitor medicine stock levels.
### Rules
#### Stock < 10 → Low Stock Alert
#### Stock ≥ 10 → Stock Sufficient

```python
## function for Medicion stock monitor
def medicine_stock_monitor(stock):
    if stock < 10:
        stock = "Low Stock Alert"
    else:
        stock = "Stock Sufficient"

    return stock

# Function call
stock = 68
status = medicine_stock_monitor(stock)
print(f"Medicine Stock: {stock}")
print(f"Status: {status}")
```
